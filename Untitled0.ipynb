```python
import shutil
from pathlib import Path
from concurrent.futures import ThreadPoolExecutor
from threading import Lock
from tqdm.auto import tqdm

FIX = Path('/content/drive/MyDrive/Satria Data 3.0/Bahan_Training_Fix')
STG = Path('/content/kaggle_upload'); STG.mkdir(exist_ok=True)

d1 = STG / 'fw-papua-train'
d2 = STG / 'fw-papua-valtest'

jobs = []
for item in ['train', 'train_rajaampat']:
    src_dir = FIX / item
    if src_dir.exists():
        for f in src_dir.rglob('*'):
            if f.is_file():
                jobs.append((f, d1 / f.relative_to(FIX)))
for j in ['class_weights.json', 'patch_sampler_weights_shared.json']:
    jobs.append((FIX / j, d1 / j))
for item in ['val', 'test']:
    for f in (FIX / item).rglob('*'):
        if f.is_file():
            jobs.append((f, d2 / f.relative_to(FIX)))

pending = []
done_bytes = 0
for src, dst in jobs:
    src_size = src.stat().st_size
    if dst.exists() and dst.stat().st_size == src_size:
        done_bytes += src_size
    else:
        pending.append((src, dst, src_size))

total_bytes = done_bytes + sum(s for _, _, s in pending)
print(f"Total {len(jobs)} file, {total_bytes/1e9:.1f} GB | sudah selesai: {done_bytes/1e9:.1f} GB | sisa: {len(pending)} file")

bar = tqdm(total=total_bytes, initial=done_bytes, unit='B', unit_scale=True, unit_divisor=1024, desc="Copy Drive -> lokal")
lock = Lock()

def _copy_one(src, dst):
    dst.parent.mkdir(parents=True, exist_ok=True)
    tmp = dst.with_suffix(dst.suffix + '.partial')
    with open(src, 'rb') as fsrc, open(tmp, 'wb') as fdst:
        while True:
            chunk = fsrc.read(8 * 1024 * 1024)
            if not chunk:
                break
            fdst.write(chunk)
            with lock:
                bar.update(len(chunk))
    tmp.rename(dst)

with ThreadPoolExecutor(max_workers=4) as exe:
    futures = [exe.submit(_copy_one, s, d) for s, d, _ in pending]
    for fut in futures:
        fut.result()

bar.close()
print('Dataset 1 (fw-papua-train) siap:', d1)
print('Dataset 2 (fw-papua-valtest) siap:', d2)
```
